#### Algún modelo con relación

```python
# If needed in Colab, first ensure Excel writer is available:
# !pip -q install openpyxl

from pathlib import Path
import pandas as pd

# === Configure your folders ===
# Example: put your CSVs in /content/input and outputs to /content/output
INPUT_DIR  = Path("/content/")
OUTPUT_DIR = Path("/content/output")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Columns to check for "No Relationship"
REL_COLS = ["rel", "rel_flanT5", "rel_robertaL", "rel_bart", "rel_deberta"]

# Final column order for the Excel export
EXPORT_COLS = ["SDGarg1", "SDGarg2", "g1", "g2", "id_arg1", "id_arg2", "rel"]

def norm(val):
    # normalize text: handle NaNs, non-breaking spaces, trim, lowercase
    s = "" if pd.isna(val) else str(val)
    return s.replace("\xa0", " ").strip().lower()

for csv_path in sorted(INPUT_DIR.glob("*.csv")):
    try:
        # Read as string to avoid accidental type coercion (e.g., "0_61")
        df = pd.read_csv(csv_path, dtype=str, keep_default_na=False)

        # Make sure required columns exist
        missing_rel = [c for c in REL_COLS if c not in df.columns]
        missing_exp = [c for c in EXPORT_COLS if c not in df.columns]
        if missing_rel:
            print(f"SKIP {csv_path.name}: missing rel columns {missing_rel}")
            continue
        if missing_exp:
            print(f"WARNING {csv_path.name}: missing export columns {missing_exp}. "
                  f"Only existing ones will be exported.")

        # Build mask: True where ALL rel* columns == "no relationship" (after normalization)
        rel_norm = df[REL_COLS].applymap(norm)
        all_no_rel = rel_norm.eq("no relationship").all(axis=1)

        # Keep rows where at least one rel* column is NOT "No Relationship"
        df_keep = df.loc[~all_no_rel].copy()

        # Select & order export columns (skip ones that are truly missing)
        cols_to_export = [c for c in EXPORT_COLS if c in df_keep.columns]
        df_out = df_keep[cols_to_export]

        # Write to Excel with same base name
        out_path = OUTPUT_DIR / f"{csv_path.stem}.xlsx"
        df_out.to_excel(out_path, index=False)  # uses openpyxl by default

        print(f"Saved {out_path.name}: {len(df_out)} rows")

    except Exception as e:
        print(f"ERROR processing {csv_path.name}: {e}")
```

## Extracción

```python
# !pip -q install scikit-learn openpyxl

from pathlib import Path
import json
import re
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# ========= CONFIG =========
INPUT_DIR  = Path("/content/Extraccion")     # <- your JSONs here
OUTPUT_DIR = Path("/content/Extraccion/out") # <- Excels will go here
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

SIM_THRESHOLD = 0.25  # add a model arg only if max cosine < 0.25 vs kept set

# ========= HELPERS =========
def dataset_prefix(p: Path) -> str:
    """
    Group key: token before the first `__`
    e.g., GLOBAL_SGD2024__AllArgs_gemma3-27b.json -> GLOBAL_SGD2024
    """
    return p.stem.split("__", 1)[0]

def group_label(prefix: str) -> str:
    """Label for the Prefix column in Excel (with trailing underscore)."""
    return f"{prefix}_"

def is_annotations_file(p: Path) -> bool:
    return "annotations" in p.name.lower()

def normalize_text(s: str) -> str:
    if s is None:
        return ""
    s = str(s).replace("\xa0", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

def max_cosine_to_set(cand: str, corpus: list[str]) -> float:
    """Return max cosine(cand, each item in corpus) using TF-IDF; 0.0 if corpus is empty."""
    if not corpus:
        return 0.0
    texts = corpus + [cand]
    vec   = TfidfVectorizer().fit_transform(texts)
    sims  = cosine_similarity(vec[-1], vec[:-1]).ravel()
    return float(sims.max()) if sims.size else 0.0

# ========= LOAD & GROUP FILES =========
json_files = sorted(INPUT_DIR.glob("*.json"))
if not json_files:
    print(f"No JSON files found in {INPUT_DIR}")

groups: dict[str, list[Path]] = {}
for p in json_files:
    groups.setdefault(dataset_prefix(p), []).append(p)

# ========= PROCESS EACH PREFIX (KW + noKW MERGED) =========
for prefix, files in groups.items():
    # Process annotations first, then other models (kw and _noKW together)
    files = sorted(files, key=lambda p: (not is_annotations_file(p), p.name))

    kept = []        # list of dicts: {"text": str, "page": int|None, "ods_set": set[int]}
    kept_texts = []  # plain texts for cosine checks
    exact_index = {} # normalized text -> kept index (for exact dupes only)

    for fp in files:
        # Per-file counters (optional for sanity checks)
        added_annot = merged_annot = added_model = merged_model = 0

        try:
            with open(fp, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"ERROR reading {fp.name}: {e}")
            continue

        if not isinstance(data, list):
            print(f"SKIP {fp.name}: top-level JSON is not a list")
            continue

        for entry in data:
            ods   = entry.get("ods")
            page  = entry.get("page")
            args  = entry.get("arguments", [])

            if not isinstance(args, list):
                continue

            for arg in args:
                text = normalize_text(arg)
                if not text:
                    continue

                if is_annotations_file(fp):
                    # Keep ALL annotations; only dedupe on exact normalized text
                    if text in exact_index:
                        idx = exact_index[text]
                        if isinstance(ods, int):
                            kept[idx]["ods_set"].add(ods)
                        if page is not None and kept[idx]["page"] is not None:
                            kept[idx]["page"] = min(kept[idx]["page"], page)
                        merged_annot += 1
                    else:
                        kept.append({
                            "text": text,
                            "page": page,
                            "ods_set": set([ods]) if isinstance(ods, int) else set()
                        })
                        kept_texts.append(text)
                        exact_index[text] = len(kept_texts) - 1
                        added_annot += 1
                else:
                    # Model files (KW and _noKW): add only if sufficiently different
                    sim = max_cosine_to_set(text, kept_texts)
                    if sim < SIM_THRESHOLD:
                        kept.append({
                            "text": text,
                            "page": page,
                            "ods_set": set([ods]) if isinstance(ods, int) else set()
                        })
                        kept_texts.append(text)
                        # allow later exact merges to find this quickly
                        exact_index.setdefault(text, len(kept_texts) - 1)
                        added_model += 1
                    else:
                        # Merge into the most similar kept item (union ODS, keep smallest page)
                        texts = kept_texts + [text]
                        vec   = TfidfVectorizer().fit_transform(texts)
                        sims  = cosine_similarity(vec[-1], vec[:-1]).ravel()
                        idx   = int(sims.argmax())
                        if isinstance(ods, int):
                            kept[idx]["ods_set"].add(ods)
                        if page is not None and kept[idx]["page"] is not None:
                            kept[idx]["page"] = min(kept[idx]["page"], page)
                        merged_model += 1

        # Optional debug log per file
        print(f"{fp.name}: +ann={added_annot}, =ann(dup)={merged_annot}, "
              f"+model={added_model}, =model(merged)={merged_model}")

    # ======= WRITE ONE EXCEL FOR THIS PREFIX =======
    rows = []
    for item in kept:
        ods_list = sorted(item["ods_set"]) if item["ods_set"] else []
        rows.append({
            "Prefix": group_label(prefix),                 # e.g., GLOBAL_SGD2024_
            "Page": item["page"],
            "ODS (0-17)": ", ".join(map(str, ods_list)) if ods_list else "",
            "Argument": item["text"],
        })

    df = pd.DataFrame(rows).sort_values(by=["Page", "Argument"], kind="stable")
    out_path = OUTPUT_DIR / f"{prefix}.xlsx"               # ONE file per prefix (KW + noKW merged)
    df.to_excel(out_path, index=False)
    print(f"Saved {out_path} with {len(df)} rows")
```
